## Explorando el dataset

```python
import pandas as pd

ruta_productos = "../data/raw/shopping_queries_dataset_products.parquet"
ruta_ejemplos = "../data/raw/shopping_queries_dataset_examples.parquet"

productos = pd.read_parquet(ruta_productos)
ejemplos = pd.read_parquet(ruta_ejemplos)

print("Productos:", productos.shape)
print("Ejemplos:", ejemplos.shape)
```

```python
print("Columnas de productos:")
print(productos.columns.tolist())

print("\nColumnas de ejemplos:")
print(ejemplos.columns.tolist())
```

```python
display(productos.head(3))
display(ejemplos.head(5))
```

```python
query_id_ejemplo = ejemplos["query_id"].iloc[0]

grupo = ejemplos[ejemplos["query_id"] == query_id_ejemplo]

display(
    grupo[[
        "query_id",
        "query",
        "product_id",
        "esci_label",
        "split"
    ]]
)
```

```python
consulta_productos = ejemplos.merge(
    productos,
    how="left",
    on=["product_locale", "product_id"]
)

display(
    consulta_productos[[
        "query",
        "product_id",
        "product_title",
        "esci_label"
    ]].head(10)
)
```

```python
print(ejemplos["product_locale"].value_counts(dropna=False))
print(ejemplos["esci_label"].value_counts(dropna=False))
print(ejemplos["split"].value_counts(dropna=False))
```
